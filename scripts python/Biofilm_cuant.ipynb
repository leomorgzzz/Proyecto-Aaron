```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
data = {
    'Cepa': ['E. coli'] * 3 + ['P. comm (WT)'] * 3 + ['P. comm (Δswap1)'] * 3,
    'Absorbancia': [2.4, 2.6, 2.5,   # E. coli (Altos)
                    0.15, 0.25, 0.2, # WT (Bajos)
                    1.7, 1.9, 1.8]   # Delta Swap
}
df = pd.DataFrame(data)
```

```python
sns.set(style="whitegrid", context="paper", font_scale=1.2)
plt.figure(figsize=(7, 6))

colors = {'E. coli': '#d62728', 'P. comm (WT)': '#1f77b4', 'P. comm (Δswap1)': '#ff7f0e'}

ax = sns.barplot(x='Cepa', y='Absorbancia', data=df, palette=colors,
                 capsize=0.1, edgecolor='black', linewidth=1.5, alpha=0.8, errwidth=1.5)

sns.stripplot(x='Cepa', y='Absorbancia', data=df,
              color='black', size=8, jitter=0.1, alpha=0.7)

plt.plot([0, 0, 1, 1], [2.7, 2.8, 2.8, 0.4], lw=1.5, c='black')
plt.text(0.5, 2.85, '****', ha='center', va='bottom', fontsize=14, fontweight='bold')

plt.plot([1, 1, 2, 2], [0.4, 2.1, 2.1, 2.0], lw=1.5, c='black')
plt.text(1.5, 2.15, '****', ha='center', va='bottom', fontsize=14, fontweight='bold')

plt.title('Figura 5B. Cuantificación de Biofilm (Cristal Violeta)', fontsize=14, fontweight='bold', pad=15)
plt.ylabel('Absorbancia ($DO_{590}$)', fontsize=12, fontweight='bold')
plt.xlabel('', fontsize=12)
plt.ylim(0, 3.2)

plt.tight_layout()
plt.savefig('Figura5B_BarDotPlot.png', dpi=300)
plt.show()
```
