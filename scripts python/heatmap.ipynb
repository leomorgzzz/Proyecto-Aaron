```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
data = {
    'Gen': ['luxS', 'lsrA', 'lsrC', 'lsrK', 'csgD', 'csgA', 'pgaA', 'pgaC', 'hfq', 'bolA'],
    'E. coli (K-12)':       [ 2.5,  2.8,  2.4,  2.9,  3.1,  3.5,  2.2,  2.6,  1.5,  1.2], # Control Positivo
    'P. comm. (WT)':        [-2.5, -2.9, -2.1, -3.0, -4.2, -3.8, -2.5, -3.1, -0.5, -0.2], # Reprimido (SWAP1 activo)
    'P. comm. (Δswap1)':    [ 2.1,  2.4,  1.9,  2.5,  2.8,  3.1,  1.8,  2.3,  1.1,  0.9]  # Rescate (SWAP1 borrado)
}
```

```python
df = pd.read_json(pd.Series(data).to_json(), orient='index').transpose()
df = pd.DataFrame(data)
df.set_index('Gen', inplace=True)
```

```python
plt.figure(figsize=(8, 6))
sns.set(font_scale=1.2)

ax = sns.heatmap(df, annot=True, cmap='vlag', center=0,
                 linewidths=.5, linecolor='white', fmt=".1f",
                 cbar_kws={'label': 'Expresión Relativa (Z-score)'})

plt.title('Figura 4B. Análisis de Expresión Diferencial (Regulón QS)', pad=20, fontsize=14, fontweight='bold')
plt.ylabel('Genes de Quorum Sensing y Biofilm', fontweight='bold')
plt.xlabel('Cepas Experimentales', fontweight='bold')


ax.xaxis.tick_top()
ax.xaxis.set_label_position('top')


plt.tight_layout()
plt.savefig('Figura4B_Heatmap.png', dpi=300)
plt.show()
```
