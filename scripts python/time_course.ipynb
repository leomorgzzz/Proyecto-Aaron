```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
tiempos = [0, 15, 30, 60, 120, 300]

wt_mean = [5, 25, 65, 95, 98, 98]
wt_err  = [2, 5,  8,  3,  2,  1]

ecoli_mean = [5, 8, 12, 20, 40, 85]
ecoli_err  = [1, 2, 3,  4,  5,  4]

mut_mean = [5, 5, 6, 5, 6, 7]
mut_err  = [2, 2, 1, 2, 1, 2]
```

```python
sns.set(style="whitegrid")
plt.figure(figsize=(8, 6))

plt.errorbar(tiempos, wt_mean, yerr=wt_err, fmt='-o', color='#1f77b4',
             linewidth=2, capsize=5, markersize=8, label='P. comm. WT')

plt.errorbar(tiempos, ecoli_mean, yerr=ecoli_err, fmt='--s', color='#d62728',
             linewidth=2, capsize=5, markersize=8, label='E. coli (Control)')

plt.errorbar(tiempos, mut_mean, yerr=mut_err, fmt='-.^', color='#555555',
             linewidth=2, capsize=5, markersize=8, label='Mutante S26A')


plt.title('Figura 6B. Cuantificación de RpoS Citoplasmático (Time-Course)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Tiempo post-fijación (segundos)', fontsize=12, fontweight='bold')
plt.ylabel('% de Fluorescencia Citoplasmática\n(Normalizada)', fontsize=12, fontweight='bold')

plt.axvspan(0, 60, color='#1f77b4', alpha=0.1, label='Ventana de Respuesta Rápida')

plt.legend(fontsize=10, loc='lower right', frameon=True, shadow=True)
plt.ylim(0, 110)
plt.xlim(-10, 320)

plt.tight_layout()
plt.savefig('Figura6B_Discrete.png', dpi=300)
plt.show()
```
