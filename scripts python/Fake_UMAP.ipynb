```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from sklearn.datasets import make_blobs
```

```python
X, y = make_blobs(n_samples=15, centers=[[2, 2], [10, 10], [3, 3]], cluster_std=0.8, random_state=42)
```

```python
df = pd.DataFrame(X, columns=['UMAP_1', 'UMAP_2'])
df['Condición'] = y
df['Condición'] = df['Condición'].replace({
    0: 'E. coli (K-12)',
    1: 'P. comm. (WT)',
    2: 'P. comm. (Δswap1)'
})
```

```python
sns.set(style="whitegrid")
plt.figure(figsize=(10, 8))
```

```python
colors = {'E. coli (K-12)': '#d62728',      # Rojo (Control)
          'P. comm. (WT)': '#1f77b4',       # Azul (Reprimido)
          'P. comm. (Δswap1)': '#ff7f0e'}   # Naranja/Rojo claro (KO)

sns.scatterplot(data=df, x='UMAP_1', y='UMAP_2', hue='Condición',
                palette=colors, s=200, alpha=0.8, edgecolor='black')

plt.title('Figura 4A. Análisis de Dimensionalidad (UMAP) del Transcriptoma', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('UMAP Dimensión 1', fontsize=12, fontweight='bold')
plt.ylabel('UMAP Dimensión 2', fontsize=12, fontweight='bold')
plt.legend(title='Cepa / Condición', title_fontsize='12', fontsize='11', loc='upper left', frameon=True, shadow=True)

plt.tight_layout()
plt.savefig('Figura4A_UMAP.png', dpi=300)
plt.show()
```
