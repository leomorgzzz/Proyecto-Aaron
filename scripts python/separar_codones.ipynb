```python
#programa para contar el numero de "*" en un archivo de texto
def contar_asteriscos(nombre_archivo):
    with open(nombre_archivo, "r") as archivo:
        contenido = archivo.read()
    return contenido.count("*")

if __name__ == "__main__":
    nombre_archivo = "secuencia_codones.txt"
    numero_asteriscos = contar_asteriscos(nombre_archivo)
    print(f"Número de '*' en el archivo '{nombre_archivo}': {numero_asteriscos}")
```

```python
#Elimina los " " de una secuencia y genera un nuevo archivo.txt y otro que elimine "*"
import textwrap
def eliminar_espacios(nombre_archivo_entrada, nombre_archivo_salida):
    with open(nombre_archivo_entrada, "r") as archivo_entrada:
        contenido = archivo_entrada.read()
    contenido_sin_espacios = contenido.replace(" ", "")
    with open(nombre_archivo_salida, "w") as archivo_salida:
        archivo_salida.write(contenido_sin_espacios)

def eliminar_asteriscos(nombre_archivo_entrada, nombre_archivo_salida):
    with open(nombre_archivo_entrada, "r") as archivo_entrada:
        contenido = archivo_entrada.read()
    contenido_sin_asteriscos = contenido.replace("*","")
    with open(nombre_archivo_salida, "w") as archivo_salida:
        archivo_salida.write(contenido_sin_asteriscos)

#funcion que lea 70 caracteres y agregue un salto de linea
def formato(nombre_archivo_entrada, nombre_archivo_salida):
    with open(nombre_archivo_entrada, "r") as archivo_entrada:
        contenido = archivo_entrada.read()
    contenido_formato = textwrap.fill(contenido, width=70)
    with open(nombre_archivo_salida, "w") as archivo_salida:
        archivo_salida.write(contenido_formato)

def distancia_hamming(nombre_archivo_entrada, nombre_archivo_salida):
    with open(nombre_archivo_entrada, "r") as archivo_entrada:
        contenido_entrada = archivo_entrada.read()
    with open(nombre_archivo_salida, "r") as archivo_salida:
        contenido_salida = archivo_salida.read()
    if len(contenido_entrada) != len(contenido_salida):
        raise ValueError("Las secuencias deben tener la misma longitud")
    return sum(c1 != c2 for c1,c2 in zip(contenido_entrada, contenido_salida))
    

    


if __name__ == "__main__":
    with open("rpoS_e.txt", "r") as f:
        contenido_rpoS = f.read()
    
    nucleotidos_rpos = len(contenido_rpoS)
    nombre_archivo = "secuencia_codones.txt"
    nombre_archivo_out = "secuencia_codones_limpia.txt"
    nombre_archivo_out_out = "secuencia_codones_limpia_limpia.txt"
    eliminar_asteriscos(nombre_archivo, nombre_archivo_out)
    eliminar_espacios(nombre_archivo_out, nombre_archivo_out_out)
    formato(nombre_archivo_out_out, "secuencia_codones_final.txt")
    distancia = distancia_hamming("secuencia_codones_final.txt", "rpoS_e.txt")
    print(f"Distancia de Hamming: {distancia}")
    distancia_int = int(distancia)
    procentaje= distancia_int/((nucleotidos_rpos)/3)*100
    print(f"Porcentaje de similitud: {procentaje}%")
```
