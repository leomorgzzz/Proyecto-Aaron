```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from scipy.stats import norm
```

```python
plt.rcParams['font.family'] = 'sans-serif'
```

```python
def plot_growth():
    fig, ax = plt.subplots(figsize=(8, 5))

    t = np.linspace(0, 12, 100)
    ecoli = 2.0 / (1 + np.exp(-1.5 * (t - 3)))
    pcomm = 0.8 / (1 + np.exp(-0.5 * (t - 6)))

    ax.plot(t, ecoli, color='#d62728', lw=3, label='E. coli (LB, 37°C)')
    ax.plot(t, pcomm, color='#1f77b4', lw=3, label='P. commutabilis (M9, 58°C)')

    ax.set_title('Figura S1. Cinética de Crecimiento Comparativa', fontweight='bold')
    ax.set_xlabel('Tiempo (Horas)')
    ax.set_ylabel('Densidad Óptica (600 nm)')
    ax.legend()
    ax.grid(True, alpha=0.3)

    ax.axvline(x=8, color='gray', linestyle='--')
    ax.text(8.2, 0.1, 'Punto de Cosecha\n(RNA-Seq)', fontsize=9)

    plt.tight_layout()
    plt.savefig('FiguraS1_Growth.png', dpi=300)
    plt.show()
```

```python
def plot_bioanalyzer():
    fig, ax = plt.subplots(figsize=(8, 4))

    x = np.linspace(0, 60, 500)
    y = 10 * np.exp(-(x-2)**2 / 0.5)
    y += 40 * np.exp(-(x-38)**2 / 2)
    y += 65 * np.exp(-(x-45)**2 / 2)
    y += np.random.normal(0, 0.5, 500)
    y = np.clip(y, 0, None)

    ax.plot(x, y, color='#2ca02c')
    ax.fill_between(x, y, color='#2ca02c', alpha=0.1)

    ax.set_title('Figura S2. Integridad del ARN Total (Bioanalyzer)', fontweight='bold')
    ax.set_xlabel('Tiempo de Migración (s)')
    ax.set_ylabel('Unidades de Fluorescencia (FU)')
    ax.set_yticks([])


    ax.text(35, 43, '16S rRNA', ha='center')
    ax.text(40, 65, '23S rRNA', ha='center')
    ax.text(55, 60, 'RIN: 9.2', fontsize=12, bbox=dict(facecolor='white', alpha=0.8))

    plt.tight_layout()
    plt.savefig('FiguraS2_RIN.png', dpi=300)
    plt.show()
```

```python
plot_growth()
```

```python
plot_bioanalyzer()
```
